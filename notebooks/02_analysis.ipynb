# 02 Analysis

Displays the tables and figures. All logic is in `src/analysis.py` and `scripts/make_figures.py`; this notebook only calls the two commands and shows what they wrote.

Run from the repo root (or set `RESULTS_DIR`). It needs no GPU, no run folders and no data.

While runs are missing it uses `--allow-partial`, so every output it shows is stamped PARTIAL. **PARTIAL numbers must not go into the report or README.**

```python
import subprocess
import sys

import pandas as pd
from IPython.display import Image, Markdown, display

from scripts.make_figures import FIGURES, SAMPLE_GRID
from src.analysis import expected_run_ids, missing_runs, output_name
from src.utils import env_path

RESULTS_DIR = env_path("RESULTS_DIR")
missing = missing_runs(RESULTS_DIR / "raw")
PARTIAL = bool(missing)
FLAGS = ["--allow-partial"] if PARTIAL else []
print(f"{len(expected_run_ids()) - len(missing)} of {len(expected_run_ids())} runs present")
print("Missing:", ", ".join(missing) or "none")
```

## Build the tables and figures

```python
for module in ("src.analysis", "scripts.make_figures"):
    subprocess.run([sys.executable, "-m", module, *FLAGS], check=True)
```

## Main results, seen and unseen, and the data for RQ1-RQ5

```python
tables = RESULTS_DIR / "tables"
for stem in ("main_results", "seen_unseen", "rq_answers"):
    display(Markdown((tables / output_name(stem, "md", PARTIAL)).read_text()))
```

## Detail tables (fractions, full precision)

```python
for stem in ("per_corruption", "per_severity", "per_family"):
    print(stem)
    display(pd.read_csv(tables / output_name(stem, "csv", PARTIAL)))
```

## Figures

```python
figures = RESULTS_DIR / "figures"
stems = sorted([*FIGURES, SAMPLE_GRID])
for stem in stems:
    partial = PARTIAL and stem != SAMPLE_GRID  # figure 7 shows no results
    print(stem)
    display(Image(figures / output_name(stem, "png", partial), width=900))
```
