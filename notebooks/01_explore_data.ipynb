# 01 Explore the data

Section 1: clean CIFAR-10. Ten training-split images per class and the class counts of the fixed train / validation / test splits (DECISIONS D1). All data access goes through `src/data/cifar10.py`; this notebook only displays the results.

Run from the repo root (or set `DATA_DIR`).

```python
import matplotlib.pyplot as plt
import pandas as pd

from src.constants import CLASS_NAMES
from src.data.cifar10 import examples_per_class, get_normalization_stats, split_class_counts
from src.utils import env_path

DATA_DIR = str(env_path("DATA_DIR"))
```

## 1.1 Ten images per class (training split)

```python
images = examples_per_class(DATA_DIR, n_per_class=10)
fig, axes = plt.subplots(10, 10, figsize=(10, 11))
for c, row in enumerate(axes):
    for j, ax in enumerate(row):
        ax.imshow(images[c, j])
        ax.axis("off")
    row[0].set_title(CLASS_NAMES[c], loc="left", fontsize=9)
plt.tight_layout()
plt.show()
```

## 1.2 Class counts per split

```python
counts = pd.DataFrame(split_class_counts(DATA_DIR), index=list(CLASS_NAMES))
counts.loc["total"] = counts.sum()
counts
```

## 1.3 Normalisation statistics (45,000 training images, D2)

```python
mean, std = get_normalization_stats()
pd.DataFrame({"mean": mean, "std": std}, index=["R", "G", "B"])
```
